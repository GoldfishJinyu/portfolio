## Popcorn Hack 3

```python
# Popcorn Hack 3 - SFI Search Filter
query = "1.2"
records = [
    {
        "product_name": "Replacement Flywheels and Clutch Assemblies",
        "category": "Auto Racing",
        "spec_number": "1.1",
    },
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Drag Racing",
        "spec_number": "1.2",
    },
    {
        "product_name": "Racing Flywheel Record",
        "category": "Auto Racing",
        "spec_number": "2.1",
    },
]
valid_categories = ["Auto Racing", "Drag Racing"]

matching_records = []

for record in records:
    # 类别必须合法
    is_valid_category = record.get("category") in valid_categories

    # 查询词匹配名称或 Spec 编号
    matches_query = (query in record.get("product_name", "")) or (
        query == record.get("spec_number", "")
    )

    if is_valid_category and matches_query:
        matching_records.append(record)

print("Matching records found:", matching_records)
```

## Homework Hack

```python
# Homework Hack - SFI Backend Validator
existing_spec_numbers = ["1.1", "2.1"]
valid_categories = ["Auto Racing", "Drag Racing"]

test_records = [
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Auto Racing",
        "spec_number": "1.2",
        "effective_date": "Feb. 9, 2006",
    },
    {
        "product_name": "Replacement Flywheels",
        "category": "Street Car",
        "spec_number": "3.1",
        "effective_date": "Jan. 1, 2026",
    },
    {
        "product_name": "Existing Flywheel Record",
        "category": "Drag Racing",
        "spec_number": "1.1",
        "effective_date": "Jan. 1, 2026",
    },
]


def validate_sfi_record(record, valid_cats, existing_specs):
    
    has_product_name = bool(record.get("product_name"))
    is_valid_category = record.get("category") in valid_cats
    has_spec_number = bool(record.get("spec_number"))
    has_effective_date = bool(record.get("effective_date"))
    is_duplicate_spec = record.get("spec_number") in existing_specs

    is_valid = (
        has_product_name
        and is_valid_category
        and has_spec_number
        and has_effective_date
        and not is_duplicate_spec
    )

    return is_valid

for idx, rec in enumerate(test_records, start=1):
    status = validate_sfi_record(rec, valid_categories, existing_spec_numbers)
    result_str = "ACCEPTED" if status else "REJECTED"
    print(f"Record {idx} ({rec['product_name']}): {result_str}")
```
