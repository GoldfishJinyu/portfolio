## Popcorn Hack 2

```python
# CODE_RUNNER: Use if/elif conditions to classify message risk
risk_score = 0
for risk_score in range(10):
    print(f"the current risk score is: {risk_score}")
    if risk_score >= 9:
        print("Do not reply. This is almost certainly a scam.")
    elif risk_score >= 6:
        print("This looks risky. Check with someone you trust before replying.")
    elif risk_score >= 3:
        print("Be a little careful, but this is probably fine.")
    else:
        print("Nothing suspicious in this message.")
```

## Homework Hack

```python
# CODE_RUNNER: 3.6 Homework - score a message, then advise on it

message = "URGENT: send gift cards now to unlock your account"
amount_requested = 500
known_contact = False

risk_score = 0

# TODO 1: add to risk_score using conditionals, one signal at a time
# Note: Independent 'if' statements are used here because each risk factor is independent 
# and cumulative. We do not use 'else' because if a signal is absent, we simply add +0 points.

if "gift card" in message:
    risk_score += 4

if amount_requested > 100:
    risk_score += 3

if not known_contact:
    risk_score += 2

# TODO 2 & 3: report on the final score with an if / elif / else chain
# 3 thresholds -> 4 outcomes -> 4 answers
if risk_score >= 8:
    advice = "CRITICAL RISK: Do not reply. This message exhibits multiple strong scam indicators."
elif risk_score >= 5:
    advice = "HIGH RISK: Proceed with extreme caution. Contact the sender through a verified external method."
elif risk_score >= 2:
    advice = "MODERATE RISK: Minor risk factors detected. Verify the sender's identity before acting."
else:
    advice = "LOW RISK: No significant risk indicators detected in this message."

print(f"Calculated Risk Score: {risk_score}")
print(f"Advice: {advice}")
```
