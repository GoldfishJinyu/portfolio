## Popcorn Hack 1

```python
# CODE_RUNNER: Traverse a list to calculate the total and average hours
# Step 1: make the hours list
hours = [3, 5, 2, 4]

# Step 2: start total at 0
total = 0

# Step 3: FOR EACH -> for ... in ...: add each h to total
for h in hours:
    total = total + h

# Step 4: LENGTH(hours) -> len(hours): compute the average
average = total / len(hours)

# Step 5: print total, then average
print(total)
print(average)
```

## Popcorn Hack 3

```python
# CODE_RUNNER: Find workshops with open seats
workshops = ["Self-Defense", "Parenting Skills", "Fitness", "Self-Sufficiency"]
signed_up = [12, 7, 15, 9]  # signed_up[0] goes with workshops[0], and so on
capacity = 12

open_workshops = []  # COLLECT names instead of counting

for i in range(len(workshops)):  # i = 0, 1, 2, 3
    if signed_up[i] < capacity:
        open_workshops.append(workshops[i])

print("Workshops with open seats:", len(open_workshops))
print(open_workshops)
```

## Homework

```python
# CODE_RUNNER: 3.09 Homework - Safe Passage Heals Volunteer & Restock Manager
# Algorithm Combining: Traversal, Accumulation, Selection, and Min/Max Finding

volunteer_hours = [4, 8, 2, 6, 12, 5]
volunteer_names = ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"]
target_threshold = 5

total_hours = 0
high_volunteers = []
max_hours = volunteer_hours[0]
top_volunteer = volunteer_names[0]

# Single-pass loop combining accumulation, collection, and max-finding
for i in range(len(volunteer_hours)):
    hours = volunteer_hours[i]
    name = volunteer_names[i]

    # 1. Accumulate total hours
    total_hours += hours

    # 2. Collect names of volunteers who met or exceeded the threshold
    if hours >= target_threshold:
        high_volunteers.append(name)

    # 3. Track the volunteer with the most hours
    if hours > max_hours:
        max_hours = hours
        top_volunteer = name

avg_hours = total_hours / len(volunteer_hours)

# Output Summary Report
print(f"Total Volunteer Hours: {total_hours}")
print(f"Average Hours per Volunteer: {avg_hours:.2f}")
print(f"Volunteers meeting threshold ({target_threshold}+ hrs): {high_volunteers}")
print(f"Top Contributor: {top_volunteer} ({max_hours} hrs)")
```
