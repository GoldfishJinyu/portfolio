## Popcorn Hack

```python
# CODE_RUNNER: Popcorn Hack - wildfire dispatch scanner
# Theme: The SFSRC team scans wildfire alert levels from different zones.
# Goal: skip safe zones, count response zones, total their alert scores, and stop at the first critical zone.

alerts = [1, 4, 2, 6, 3, 9, 5]

# Pick the cutoff for safe alerts. Alerts below this number should be skipped.
safe_cutoff = 3

# Pick the cutoff for critical alerts. The loop should stop at this number or higher.
critical_cutoff = 8

# Counters and totals start at 0 because no alerts have been processed yet.
action_count = 0
safe_count = 0
total_action_score = 0
critical_found = False

# Loop through the alert list.
for alert in alerts:
    # If the alert is safe, count it, print a message, and skip to the next alert.
    if alert < safe_cutoff:
        safe_count += 1
        print("Alert " + str(alert) + ": safe, skipping")
        continue

    # If code reaches here, the alert needs action.
    # Count this action alert and add its score to the running total.
    action_count += 1
    total_action_score += alert
    print("Alert " + str(alert) + ": action needed")

    # If this alert is critical, mark it and stop the scan.
    if alert >= critical_cutoff:
        critical_found = True
        print("Critical alert found - stopping scan")
        break

# Final report: these lines should explain what happened during the scan.
print("Safe alerts skipped: " + str(safe_count))
print("Action alerts: " + str(action_count))
print("Total action score: " + str(total_action_score))
print("Critical found: " + str(critical_found))
```

## Howework Hack

```python
# CODE_RUNNER: Homework Hack - themed loop-based checker
# Theme: Server CPU Load Monitoring System
# Goal: skip safe values, count action values, total their scores, and stop at the first critical value.

server_loads = [12, 45, 28, 68, 35, 92, 75, 10]

# Values below safe_cutoff (30% load) are skipped.
safe_cutoff = 30

# Values at or above critical_cutoff (85% load) trigger emergency shutdown.
critical_cutoff = 85

checked_count = 0
action_count = 0
skipped_count = 0
total_score = 0
critical_found = False

# Traverse every load value in the list until a critical value stops the scan.
for value in server_loads:
    checked_count += 1
    
    # Skip safe values.
    if value < safe_cutoff:
        skipped_count += 1
        print(str(value) + "%: safe load, no action required")
        continue

    # Action required for non-safe loads.
    action_count += 1
    total_score += value
    print(str(value) + "%: high load, dispatching cooling fan")

    # Stop early if load is critical.
    if value >= critical_cutoff:
        critical_found = True
        print("CRITICAL LOAD DETECTED (" + str(value) + "%). Emergency brake initiated.")
        break

# Final report
print("\n--- Final Report ---")
print("Checked: " + str(checked_count))
print("Skipped: " + str(skipped_count))
print("Actions: " + str(action_count))
print("Total score: " + str(total_score))
print("Critical found: " + str(critical_found))
```
