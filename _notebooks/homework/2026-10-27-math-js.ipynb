## Homework Hack

```python
const itemPrice = 12;
const quantity = 4; 
const discountRate = 0.25;
const budget = 30;


const subtotal = itemPrice * quantity;
const discount = subtotal * discountRate;
const total = subtotal - discount;
const withinBudget = total <= budget;

const results = [
    `Subtotal: ${subtotal}`,
    `Discount: ${discount}`,
    `Total: ${total}`,
    `Within budget: ${withinBudget}`
];
console.log(results.join("\n"));
```
