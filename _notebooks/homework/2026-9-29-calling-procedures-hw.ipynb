## Popcorn Hack 2: Decomposition & Helper Procedures

```python
PROCEDURE is_priority(incident_type)
{
    RETURN (incident_type = "Medical Emergency")
}

PROCEDURE process_report(location, incident_type)
{
    priority ← is_priority(incident_type)
    IF (priority)
    {
        DISPLAY("PRIORITY: " + incident_type + " reported in " + location)
    }
    ELSE
    {
        DISPLAY("STANDARD REPORT: " + incident_type + " reported in " + location)
    }
}

process_report("Poway", "Medical Emergency")
process_report("Poway", "Power Outage")
```

```python
# CODE_RUNNER: Use a helper function to choose a dispatch response
# Helper Procedure
def is_priority(incident_type):
    return incident_type == "Medical Emergency"


# Outer Procedure
def process_report(location, incident_type):
    priority = is_priority(incident_type)

    if priority:
        print(f"PRIORITY: {incident_type} reported in {location}")
    else:
        print(f"STANDARD REPORT: {incident_type} reported in {location}")


# Test calls
process_report("Poway", "Medical Emergency")
process_report("Poway", "Power Outage")
```

## Homework Hack

```python
# CODE_RUNNER: Dispatch different incidents using a helper function
# Helper Procedure
def check_priority(incident_type):
    # Returns True for urgent priority incidents
    priority_incidents = ["Medical Emergency", "Structure Fire", "Severe Flood"]
    return incident_type in priority_incidents


# Outer Procedure
def dispatch_response(location, incident_type):
    # Call the helper function and store the boolean result
    is_priority = check_priority(incident_type)

    # Print conditional response based on priority
    if is_priority:
        print(
            f"EMERGENCY DISPATCH: High-priority unit dispatched for {incident_type} at {location}!"
        )
    else:
        print(
            f"ROUTINE DISPATCH: Standard team logged for {incident_type} at {location}."
        )


# Test calls (3 calls with different incident types and locations)
dispatch_response("Poway", "Medical Emergency")
dispatch_response("Rancho Bernardo", "Power Outage")
dispatch_response("San Diego", "Structure Fire")
```
