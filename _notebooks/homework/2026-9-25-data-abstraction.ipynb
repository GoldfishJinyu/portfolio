## Homework

````python
```python
# Refactored UESL Star Trail Prototype using Data Abstraction
level_stars = [12, 18, 10]

# Update second level (+5) and append a new level (7)
if len(level_stars) >= 2:
    level_stars[1] = level_stars[1] + 5

level_stars.append(7)

# Traversal loop to calculate total stars
total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars

print("Level Stars Collection:", level_stars)
print("Level Count:", len(level_stars))
print("Total Stars Collected:", total_stars)
```
````
